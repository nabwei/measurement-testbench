# Measurement uncertainty walkthrough
Synthetic length model in mm. First-order propagation is cross-checked with explicitly normal/rectangular input distributions. k=2 does not establish accreditation or exact 95% coverage.

```python
from uncertainty import type_a, rectangular, propagate, monte_carlo, Sampling
components = [type_a("length_mm", [100.01, 100.00, 100.02, 99.99, 100.01]), rectangular("zero_mm", 0, 0.02)]
def model(inputs):
    return inputs[0] + inputs[1]
gum = propagate(model, components)
gum
```

```python
simulation = monte_carlo(model, components, Sampling(100000, 2026))
relative_difference = abs(simulation["standard_uncertainty"] / gum["standard_uncertainty"] - 1)
assert relative_difference < 0.02
simulation
```
